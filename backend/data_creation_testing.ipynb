```python
import fastf1
import fastf1.core
import numpy as np
import pandas as pd
from IPython.display import display

fastf1.Cache.enable_cache('.fastf1_cache/')
```

```python
session = fastf1.get_session(2024, 'miami', 'r')
session.load(telemetry=True)
```

```python
telemetry = session.laps.pick_driver('4').get_telemetry().add_driver_ahead()

display(telemetry['Time'])

ar = []

for i in range(len(telemetry['Time'][1:])):
    q = telemetry['Time'].iloc[i+1] - telemetry['Time'].iloc[i]
    ar.append(q.total_seconds())
    
    # ar.append(().astype('timedelta64[ms]').astype('int'))
hzt = 0
for i in ar:
    print(f"time since prev: {i}")
    hzt += i

print(f"arverage Hz: {(hzt / len(ar)) ** -1}")

# l = len(telemetry)
# with open("test.csv", 'w') as f:
    # f.write(telemetry[int(l*0):int(l*1.0)].to_csv())
# display(telemetry[int(l*0.1):int(l*.9)])
```

```python
driver = '1'

for driver in session.drivers:
    print(driver)
    frequency = 10 # Hz
    delta = pd.Timedelta(seconds=(1/frequency))
    telemetry = session.laps.pick_drivers(driver).get_telemetry()
    row_index = 1
    start = pd.Timedelta(seconds=0) + 20000 * delta
    # while True:
    for _ in range(1):
        while telemetry.iloc[row_index]["Time"] < start:
            row_index += 1
        df_with_best_timestamp = telemetry.iloc[row_index - 1]

        # print(f"Current timestamp: {start}, \t\treal timestamp: {df_with_best_timestamp["Time"]}")
        display(df_with_best_timestamp)
        start += delta
```
